```python
import markdown
import pdfkit
import re
import os

def split_markdown_sections(markdown_text):
    sidebar_sections = ["Technical Skills", "Certifications", "Languages", "Interests"]
    lines = markdown_text.splitlines()
    sidebar_md = []
    main_md = []
    current_section = []
    target = main_md

    for line in lines:
        if re.match(r"^##\s+", line):
            header_text = line.replace("##", "").strip()
            if header_text in sidebar_sections:
                if current_section:
                    target.extend(current_section)
                    current_section = []
                target = sidebar_md
            else:
                if current_section:
                    target.extend(current_section)
                    current_section = []
                target = main_md
        current_section.append(line)

    if current_section:
        target.extend(current_section)

    return "\n".join(sidebar_md), "\n".join(main_md)

def convert_markdown_to_pdf_local(markdown_path, output_pdf="Resume.pdf", wkhtmltopdf_path=None):
    with open(markdown_path, "r", encoding="utf-8") as f:
        markdown_content = f.read()

    sidebar_md, main_md = split_markdown_sections(markdown_content)
    sidebar_html = markdown.markdown(sidebar_md, extensions=["extra", "smarty"])
    main_html = markdown.markdown(main_md, extensions=["extra", "smarty"])

    styled_html = f"""
    <!DOCTYPE html>
    <html lang="en">
    <head>
        <meta charset="UTF-8">
        <style>
            @page {{
                size: A4;
                margin: 0;
            }}
            body {{
                margin: 0;
                font-family: 'Arial', sans-serif;
                font-size: 11pt;
                background: #ffffff;
                color: #000;
            }}
            .resume {{
                display: flex;
                flex-direction: row;
                min-height: 297mm;
                width: 210mm;
            }}
            .sidebar {{
                width: 30%;
                background-color: #f0f0f0;
                padding: 25px 20px;
                box-sizing: border-box;
            }}
            .main {{
                width: 70%;
                padding: 30px 30px;
                box-sizing: border-box;
            }}
            .header {{
                width: 100%;
                text-align: center;
                background: #333;
                color: white;
                padding: 20px 0;
                font-size: 24pt;
                font-weight: bold;
            }}
            h2 {{
                font-size: 14pt;
                margin-top: 20px;
                border-bottom: 1px solid #ccc;
                padding-bottom: 5px;
            }}
            h3 {{
                font-size: 12pt;
                margin-top: 15px;
            }}
            p, li {{
                line-height: 1.5;
                margin: 4px 0;
            }}
            ul {{
                padding-left: 20px;
            }}
        </style>
    </head>
    <body>
        <div class="header">Kevin Andrews</div>
        <div class="resume">
            <div class="sidebar">{sidebar_html}</div>
            <div class="main">{main_html}</div>
        </div>
    </body>
    </html>
    """

    # Save and convert to PDF
    html_path = "temp_resume.html"
    with open(html_path, "w", encoding="utf-8") as f:
        f.write(styled_html)

    config = pdfkit.configuration(wkhtmltopdf=wkhtmltopdf_path) if wkhtmltopdf_path else None
    pdfkit.from_file(html_path, output_pdf, configuration=config)
    print(f"✅ PDF saved successfully at {output_pdf}")
    os.remove(html_path)

# Usage
if __name__ == "__main__":
    wkhtmltopdf_path = r"D:\Help_project_list\resume_builder\reume_builder_checking\ai-resume-creator\wkhtmltox\bin\wkhtmltopdf.exe"
    convert_markdown_to_pdf_local("optimized_resume.md", "Kevin_Andrews_Resume3.pdf", wkhtmltopdf_path)
```

```python
import requests
from bs4 import BeautifulSoup




def get_job_description_from_url(url):
    response = requests.get(url)
    if response.status_code == 200:
        soup = BeautifulSoup(response.text, "html.parser")
        # You might need to adjust the selector based on the actual structure
        job_description = soup.get_text(separator="\n", strip=True)
        return job_description
    else:
        raise Exception(f"Failed to fetch URL: {response.status_code}")


# Example usage
url = "https://jobs.six-group.com/job/Warsaw-%28Senior%29-FrontFull-stack-Developer/1193235301/"
job_text = get_job_description_from_url(url)
print(job_text)
```
